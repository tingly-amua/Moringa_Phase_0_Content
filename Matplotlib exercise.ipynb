```python
#using plt
x = [2,2,2,2,2.3,2.6,2.6,3,3.1,3.2,3.6,3.8,3.8,4]
y = [50,48,100,12,34,30,78,87,84,75,83,90,90,98]

y2 = [60,58,110,22,44,40,88,97,94,85,93,100,100,108]

plt.figure(figsize = (20,6))
plt.scatter(x,y, color = 'b', s = 200, label = 'Student Motivation')
plt.scatter(x,y2, color = 'green', s = 200, label = 'Alternate axis', marker = 's')

plt.title('Student Performance Analysis', fontsize = 20)
plt.xlabel('x-axis', fontsize = 15)
plt.ylabel('y-axis', fontsize = 15)
plt.grid(True, linestyle = '-.', alpha = 0.9, color = 'black')
plt.legend(fontsize = 20)




plt.show()
```

```python
#using ax
x = [2,2,2,2,2.3,2.6,2.6,3,3.1,3.2,3.6,3.8,3.8,4]
y = [50,48,100,12,34,30,78,87,84,75,83,90,90,98]

fig, (ax1,ax2) = plt.subplots(1,2,figsize = (10,6))
#fig, ax = plt.subplots(figsize = (20,6)) #important to define ax function

#subplot1
ax1.plot(x,y, color = 'r', label = 'Student Performance', marker = '^')

ax1.set_title('Performance Distribution1', fontsize = 20)
ax1.set_xlabel('x-axis', fontsize = 15)
ax1.set_ylabel('y-axis', fontsize = 15)
ax1.grid(True)
ax1.legend()

#subplot2
ax2.plot(x,y, color = 'blue', label = 'Student Performance Excerpt', marker = 's')

ax2.set_title('Performance Distribution2', fontsize = 20)
ax2.set_xlabel('x-axis', fontsize = 15)
ax2.set_ylabel('y-axis', fontsize = 15)
ax2.grid(True)
ax2.legend()

ax2.set_xlim(3.25,4.0)
ax2.set_ylim(60, 100)




plt.tight_layout()
plt.show()
```

```python
data1 = [2,2,2,2,2.3,2.6,2.6,3,3.1,3.2,3.6,3.8,3.8,4]
data2 = [50,48,100,12,34,30,78,87,84,75,83,90,90,98]

#calculating covariance

#means for x and y
mean_x = np.mean(x)
mean_y = np.mean(y)
#print (mean_x, mean_y)

numerator_x = [(x-mean_x) for x in data1]
print(numerator_x)

numerator_y = [(x-mean_y) for x in data1]
print(numerator_y)

```

```python
import pandas as pd
file_name = './cities.xlsx'
travel_df = pd.read_excel(file_name)
cities = travel_df.to_dict('records')
```

```python
cities
```

```python
df = pd.DataFrame(cities)
```

```python
city_list = df['City'].values.tolist()
print('City list:', city_list)
```

```python
pop_list = df['Population'].values.tolist()
print('Population:', pop_list)
```

```python
#prepare data
x = city_list
y = pop_list

#set canvas
plt.figure(figsize = (15,5))

#plot data
plt.bar(x,y, color=['green', 'red', 'yellow', 'orange', 'pink', 'purple', 'black', 'gold', 'purple', 'orange', 'violet', 'maroon'], edgecolor = 'black', linewidth = 1)
plt.title('Bar Graph of City Vs. Population Size', fontsize = 20)
plt.xlabel('City Names', fontsize = 15)
plt.ylabel('Population Size',fontsize = 20)
plt.tight_layout()
```
