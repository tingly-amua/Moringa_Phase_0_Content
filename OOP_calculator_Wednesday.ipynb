### Class creation, Constructor method, Attributes, and Methods

```python
#Create a class
class Car: #define new class named car - a class is a blueprint for creating objects
    def __init__(self, brand, color, mileage = 0): #the car has 2 attributes, brand and color #def init is a danda method #default attributes for our parameters in our constructor method such that it makes them optional when creating an object
        self.brand = brand #create an instance variable brand for that specific car and assign it the value that will be passed in
        self.color = color #every object of the car class being created will have its own brand and color
        self.mileage = mileage
    def drive(self): #the behavior this car has - drive
        print(f"The {self.color} {self.brand} is being driven!") #formatted string we have created a car class
    def describe(self):
        print(f"The {self.brand} she/he is driving is {self.color} and has {self.mileage}")
my_car = Car("Toyota","Blue") #we have created an instance - no self here, we call self when defining the method, but not here

print(my_car.color)
print(my_car.brand)

my_car.drive() #this is a method, dont forget the brackets
```

* The def init is a method in itself -  **a constructor method** -  a special method in Python automatically called when you create a new object from the class
* Self refers to the specific instance of the class being created
    * A reference to the current object being called
    * what our instance method will use to access and modify the instance's own attributes
    * always include self as the first parameter in instance methods but dont pass it manually whem calling the method
* Brand and color are the parameters passed when we create our own object
* In the above, the **Instance Attribute** is the brand and color ie what is giving the object its properties, whereas the **Instance Method** is how that Car would behave, ie the def drive bit -  a method called on an object(an instance of the class)

```python
her_car = Car("Mercedes", "Black", 100) #an instance of that car
her_car.describe()
```

```python
#Creating a diffrent class - Student
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade #the instance attributes
    def introduce(self):
        print(f"Hi, I am {self.name} and I am in grade {self.grade}.")

student1 = Student("Tingly", 10)
student2 = Student("Timothy", 11)

student1.introduce()
student2.introduce()
```

### Inheritance

```python
#Inheritance example - no attributes here so we cant manipulate
class Animal: #superclass
    def speak(self):
        print("This animal makes a sound")

class Dog(Animal): #subclass - inherited Animal characteristics, if there were attributes i think
    def speak(self):
        print("The dog barks") #overriding initial method - method overriding - produce different output

my_dog = Dog()
my_dog.speak()
```
