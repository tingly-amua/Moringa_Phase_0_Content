```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 30)  
y = np.sin(x) 
y2=np.cos(x)
plt.figure(figsize=(10,6))
plt.scatter(x,y, color="orange",label="Sin data point",marker="o",s=100)
plt.scatter(x,y2, color="g",label="cos data point",marker="s",s=200)
plt.title("Sin distribution", fontsize=16)
plt.xlabel("X-axis",fontsize=12)
plt.ylabel("Y-axis",fontsize=12)
plt.legend(loc=3)
plt.grid(True,ls="--")
plt.show()
```

```python
fruits = ['Apples', 'Bananas', 'Oranges', 'Strawberries', 'Grapes']
sales = [45, 30, 25, 40, 35]  # Sales in units 
plt.figure(figsize=(10,6))
plt.bar(fruits,sales, color=["r","orange","purple","pink","violet"],width=0.8,edgecolor="black")
plt.title("Fruits sale",fontsize=16)
plt.xlabel("Fruits",fontsize=12)
plt.ylabel("Sales units",fontsize=12)
for i, value in enumerate(sales):
    plt.text(i, value + 1, str(value), ha='center', fontsize=12)
    
    
plt.show()
```

```python
ages = np.random.normal(100, 10, 1000)
plt.figure(figsize=(8,6))
plt.hist(ages,bins=7,color="b",width=5.0,edgecolor="black")
plt.title("Sort ages by height")
plt.xlabel("ages(years)")
plt.ylabel("height")
plt.grid(True, ls="--",alpha=0.7)
plt.show()
```

```python
ages = np.random.uniform(100, 10, 1000)
plt.figure(figsize=(8,6))
plt.hist(ages,bins=7,color="b",width=5.0)
plt.title("Sort ages by height")
plt.xlabel("ages(years)")
plt.ylabel("height")
plt.show()
```

```python
fruits = ['Apples', 'Bananas', 'Oranges', 'Strawberries', 'Grapes']
sales = [45, 30, 25, 40, 35]  # Sales in units 
plt.figure(figsize=(10,6))
plt.pie(sales,labels=fruits,colors=["g","yellow","orange","r","purple"],autopct="%0.1f%%",explode=(0,0,0,0.2,0),shadow=True)
plt.title("Fruits Distribution",fontsize=18)
plt.show()
```

```python
x = np.linspace(0, 10, 30)
y1 = np.sin(x) 
y2=np.cos(x)
fig, axs=plt.subplots(2,1, figsize=(10,6))
#subplot 1
axs[0].plot(x,y1,color="r",label="sin presentation")
axs[0].set_title("Sin waves")
axs[0].set_xlabel("x-axis")
axs[0].set_ylabel("y-axis")
axs[0].legend()
axs[0].grid(True)


#subplot 2
axs[1].plot(x,y2,color="g",label="cos presentation")
axs[1].set_title("cos waves")
axs[1].set_xlabel("x-axis")
axs[1].set_ylabel("y-axis")
axs[1].legend()
axs[1].grid(True)

plt.show()
```
